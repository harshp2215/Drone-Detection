```python
import os
import random
import shutil
```

```python
SOURCE_ROOT = "datasets"
DEST_ROOT = "final_dataset"

splits = {"train": 0.7, "val": 0.2, "test": 0.1}

classes = [
    "aeroplane",
    "bird",
    "bird_drone",
    "drone",
    "missile",
    "mosquito",
    "mosquito_drone"
]
```

```python

for split in splits:
    os.makedirs(os.path.join(DEST_ROOT, "images", split), exist_ok=True)
    os.makedirs(os.path.join(DEST_ROOT, "labels", split), exist_ok=True)

for cls in classes:

    img_dir = os.path.join(SOURCE_ROOT, cls, "images")
    label_dir = os.path.join(SOURCE_ROOT, cls, "labels")

    images = [f for f in os.listdir(img_dir) if f.endswith(".jpg")]
    random.shuffle(images)

    total = len(images)
    train_end = int(total * splits["train"])
    val_end = train_end + int(total * splits["val"])

    split_data = {
        "train": images[:train_end],
        "val": images[train_end:val_end],
        "test": images[val_end:]
    }

    for split in split_data:
        for img_name in split_data[split]:

            shutil.copy(
                os.path.join(img_dir, img_name),
                os.path.join(DEST_ROOT, "images", split, img_name)
            )

            label_name = img_name.replace(".jpg", ".txt")
            shutil.copy(
                os.path.join(label_dir, label_name),
                os.path.join(DEST_ROOT, "labels", split, label_name)
            )

print("Dataset split complete!")
```
