```python
import os
import cv2
import numpy as np
import random

CUTOUT_DIR = "all_cutouts/Missile"
BACKGROUND_DIR = "background/backgrounds"

OUTPUT_IMG_DIR = "datasets/mis/images"
OUTPUT_LABEL_DIR = "datasets/mis/labels"

CLASS_ID = 4
NUM_IMAGES = 1000
IMG_SIZE = 960

os.makedirs(OUTPUT_IMG_DIR, exist_ok=True)
os.makedirs(OUTPUT_LABEL_DIR, exist_ok=True)

def add_noise(image):
    noise = np.random.normal(0, 10, image.shape).astype(np.int16)
    noisy = image.astype(np.int16) + noise
    return np.clip(noisy, 0, 255).astype(np.uint8)

def create_alpha_mask(cutout):
    if cutout.shape[2] == 4:
        alpha = cutout[:, :, 3] / 255.0
        bgr = cutout[:, :, :3]
        return bgr, alpha

    gray = cv2.cvtColor(cutout, cv2.COLOR_BGR2GRAY)
    _, mask = cv2.threshold(gray, 200, 255, cv2.THRESH_BINARY_INV)
    mask = cv2.GaussianBlur(mask, (3,3), 0)
    alpha = mask / 255.0
    return cutout, alpha

cutouts = os.listdir(CUTOUT_DIR)
backgrounds = os.listdir(BACKGROUND_DIR)

for i in range(NUM_IMAGES):

    bg_path = os.path.join(BACKGROUND_DIR, random.choice(backgrounds))
    bg = cv2.imread(bg_path)
    bg = cv2.resize(bg, (IMG_SIZE, IMG_SIZE))

    cutout_path = os.path.join(CUTOUT_DIR, random.choice(cutouts))
    cutout = cv2.imread(cutout_path, cv2.IMREAD_UNCHANGED)

    scale = random.uniform(0.3, 0.7)
    new_w = int(IMG_SIZE * scale)
    new_h = int(cutout.shape[0] * (new_w / cutout.shape[1]))
    cutout = cv2.resize(cutout, (new_w, new_h))

    bgr, alpha = create_alpha_mask(cutout)

    brightness = random.uniform(0.8, 1.2)
    bgr = cv2.convertScaleAbs(bgr, alpha=brightness, beta=0)

    x_offset = random.randint(0, IMG_SIZE - new_w)
    y_offset = random.randint(0, IMG_SIZE - new_h)

    for c in range(3):
        bg[y_offset:y_offset+new_h, x_offset:x_offset+new_w, c] = \
            alpha * bgr[:, :, c] + \
            (1 - alpha) * bg[y_offset:y_offset+new_h, x_offset:x_offset+new_w, c]

    if random.random() < 0.5:
        bg = cv2.GaussianBlur(bg, (3,3), 0)

    bg = add_noise(bg)

    img_name = f"img_{i}.jpg"
    cv2.imwrite(os.path.join(OUTPUT_IMG_DIR, img_name), bg)

    x_center = (x_offset + new_w / 2) / IMG_SIZE
    y_center = (y_offset + new_h / 2) / IMG_SIZE
    width = new_w / IMG_SIZE
    height = new_h / IMG_SIZE

    label_path = os.path.join(
        OUTPUT_LABEL_DIR,
        img_name.replace(".jpg", ".txt")
    )

    with open(label_path, "w") as f:
        f.write(f"{CLASS_ID} {x_center} {y_center} {width} {height}")

print("dataset with YOLO labels generated!")
```
